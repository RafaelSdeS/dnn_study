# PCAD Results Analysis

Unified analysis of PCAD (Performance Characterization and Analysis for DNN) large-scale training results.

## Configuration

````python
import json
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from pathlib import Path


def find_project_root(start: Path | None = None):
    """Walk upward until the repository root is found."""
    start = start or Path.cwd()
    for candidate in (start, *start.parents):
        if (candidate / "README.md").exists() and (candidate / "results").exists() and (candidate / "models").exists():
            return candidate
    return start


project_root = find_project_root()
sys.path.insert(0, str(project_root))  # so `import ml`/`import scripts` resolve regardless of kernel cwd
print(f"Project root: {project_root}")

PCAD_DIR = project_root / "outputs" / "pcad"
LARGE_SCALE_DIR = PCAD_DIR / "archive_legacy_phases" / "phase_4_5_large_scale"

# Figures output
FIGURES_DIR = project_root / "results" / "figures_generated" / "phase_9_bypass_ablation"
FIGURES_DIR.mkdir(parents=True, exist_ok=True)

from ml.plotting import apply_report_style, PALETTE, STAGE_COLORS, RED, GREEN, NEUTRAL, group_colors_for, group_legend
apply_report_style(palette=PALETTE)

print(f"PCAD directory: {PCAD_DIR}")
print(f"Output figures to: {FIGURES_DIR}")
````

## Load Data

````python
def load_pcad_comparison():
    """Load results_phase_4_5_large_scale.csv from PCAD results."""
    csv_path = project_root / "results" / "results_aggregate" / "results_phase_4_5_large_scale.csv"
    if not csv_path.exists():
        print(f"[skip] No results_phase_4_5_large_scale.csv found")
        return pd.DataFrame()
    df = pd.read_csv(csv_path)
    return df


def load_model_summaries():
    """Load all *_summary.json files from PCAD (any subdirectory)."""
    rows = []
    if not PCAD_DIR.exists():
        print(f"[skip] {PCAD_DIR} not found")
        return pd.DataFrame()

    # Search all subdirectories recursively for *_summary.json, excluding budget_unico/
    # (Winograd-FPGA sweep provenance -- different schema, no model_name field)
    for model_dir in sorted(PCAD_DIR.rglob("*_summary.json")):
        if "budget_unico" in model_dir.parts:
            continue
        try:
            with open(model_dir) as fp:
                data = json.load(fp)
            rows.append(data)
        except Exception as e:
            print(f"[skip] {model_dir}: {e}")

    df = pd.DataFrame(rows) if rows else pd.DataFrame()
    if not df.empty and "model_name" in df.columns and "epochs" in df.columns:
        # Ablation models (e.g. alexnet_fire_bypass) can have multiple summary.json files —
        # an early-stopped ablation run plus a later full-budget rerun. Keep the most-trained
        # (highest-epoch) one per model_name so charts don't get duplicate/conflicting rows.
        df = df.sort_values("epochs").drop_duplicates(subset="model_name", keep="last").reset_index(drop=True)
    return df


df_comp = load_pcad_comparison()
df_summ = load_model_summaries()

print(f"Comparison rows : {len(df_comp):3d}")
print(f"Per-model JSONs : {len(df_summ):3d}")
print(f"\nComparison columns: {list(df_comp.columns[:10])}..." if len(df_comp.columns) > 10 else f"\nComparison columns: {list(df_comp.columns)}")
````

### 3. Accuracy vs Model Size

````python
def pareto_front_mask(xs, ys):
    """Boolean mask: True if point is Pareto-optimal (minimize x, maximize y)."""
    xs, ys = np.asarray(xs, float), np.asarray(ys, float)
    dominated = np.zeros(len(xs), dtype=bool)
    for i in range(len(xs)):
        for j in range(len(xs)):
            if i != j and xs[j] <= xs[i] and ys[j] >= ys[i] and (xs[j] < xs[i] or ys[j] > ys[i]):
                dominated[i] = True
                break
    return ~dominated


if not df_data.empty and all(c in df_data.columns for c in ["fp32_top1", "fp32_size_mb"]):
    df_fp32_all = df_data[df_data["fp32_top1"].notna() & df_data["fp32_size_mb"].notna()].copy()
    df_int8_all = df_data[df_data["int8_top1"].notna() & df_data["int8_size_mb"].notna()].copy()

    # FP32 and INT8 compress very differently (e.g. resnet18tv/vgg_style only
    # lead post-quantization) -- so compute the two Pareto fronts separately.
    fp32_pf_names, int8_pf_names = set(), set()
    if len(df_fp32_all) >= 2:
        mask = pareto_front_mask(df_fp32_all["fp32_size_mb"].values, df_fp32_all["fp32_top1"].values)
        fp32_pf_names = set(df_fp32_all.loc[mask, "model_name"])
    if len(df_int8_all) >= 2:
        mask = pareto_front_mask(df_int8_all["int8_size_mb"].values, df_int8_all["int8_top1"].values)
        int8_pf_names = set(df_int8_all.loc[mask, "model_name"])

    # Declutter: keep models Pareto-optimal in FP32 or INT8, plus the Phase 9
    # bypass-ablation models regardless of dominance -- they're this notebook's subject.
    ablation_names = set(df_data.loc[df_data["model_name"].str.contains("bypass"), "model_name"])
    keep_names = fp32_pf_names | int8_pf_names | ablation_names
    df_fp32 = df_fp32_all[df_fp32_all["model_name"].isin(keep_names)].copy()
    df_int8 = df_int8_all[df_int8_all["model_name"].isin(keep_names)].copy()

    fig, ax = plt.subplots(figsize=(11, 7))

    # FP32 -> INT8 connector lines
    for _, fp32_row in df_fp32.iterrows():
        int8_match = df_int8[df_int8["model_name"] == fp32_row["model_name"]]
        if not int8_match.empty:
            int8_row = int8_match.iloc[0]
            ax.plot([fp32_row["fp32_size_mb"], int8_row["int8_size_mb"]],
                   [fp32_row["fp32_top1"], int8_row["int8_top1"]],
                   color="gray", lw=1, alpha=0.35, zorder=1)

    # FP32 scatter (circles), INT8 scatter (squares) -- colored by model family
    ax.scatter(df_fp32["fp32_size_mb"], df_fp32["fp32_top1"],
               c=group_colors_for(df_fp32["model_name"]), marker="o", s=140,
               zorder=3, edgecolors="white", lw=0.5, alpha=0.9)

    if not df_int8.empty:
        ax.scatter(df_int8["int8_size_mb"], df_int8["int8_top1"],
                   c=group_colors_for(df_int8["model_name"]), marker="s", s=140,
                   zorder=3, edgecolors="white", lw=0.5, alpha=0.9)

    # FP32 Pareto front (blue dashed step) -- tag points that drop off the INT8 front
    if len(fp32_pf_names) >= 2:
        pf = df_fp32[df_fp32["model_name"].isin(fp32_pf_names)].sort_values("fp32_size_mb")
        ax.step(pf["fp32_size_mb"], pf["fp32_top1"], where="post",
                color=COLORS["fp32"], lw=1.5, ls="--", alpha=0.8, zorder=2)
        for _, row in pf.iterrows():
            tag = "" if row["model_name"] in int8_pf_names else "  (FP32-only)"
            ax.annotate(row["model_name"] + tag, (row["fp32_size_mb"], row["fp32_top1"]),
                        xytext=(6, 6), textcoords="offset points", fontsize=8.5,
                        fontweight="bold", color=COLORS["fp32"], alpha=0.95)

    # INT8 Pareto front (red dashed step) -- tag points that drop off the FP32 front
    if len(int8_pf_names) >= 2:
        pf = df_int8[df_int8["model_name"].isin(int8_pf_names)].sort_values("int8_size_mb")
        ax.step(pf["int8_size_mb"], pf["int8_top1"], where="post",
                color=COLORS["int8"], lw=1.5, ls="--", alpha=0.8, zorder=2)
        for _, row in pf.iterrows():
            tag = "" if row["model_name"] in fp32_pf_names else "  (INT8-only)"
            ax.annotate(row["model_name"] + tag, (row["int8_size_mb"], row["int8_top1"]),
                        xytext=(6, -11), textcoords="offset points", fontsize=8.5,
                        fontweight="bold", color=COLORS["int8"], alpha=0.95)

    # Ablation models aren't on either front -- label them directly (both quant stages)
    # so they don't get lost among the unlabeled dominated points.
    for _, row in df_fp32[df_fp32["model_name"].isin(ablation_names)].iterrows():
        ax.annotate(row["model_name"], (row["fp32_size_mb"], row["fp32_top1"]),
                    xytext=(6, 6), textcoords="offset points", fontsize=8.5,
                    fontweight="bold", color=group_colors_for([row["model_name"]])[0], alpha=0.95)
    for _, row in df_int8[df_int8["model_name"].isin(ablation_names)].iterrows():
        ax.annotate(row["model_name"], (row["int8_size_mb"], row["int8_top1"]),
                    xytext=(6, -11), textcoords="offset points", fontsize=8.5,
                    fontweight="bold", color=group_colors_for([row["model_name"]])[0], alpha=0.95)

    ax.set_xscale("log")
    ax.set_xlabel("Model Size  (MB, log scale)")
    ax.set_ylabel("Top-1 Accuracy (%)")
    ax.set_title(f"PCAD: Accuracy vs Model Size -- Pareto-optimal + ablation models "
                 f"({len(keep_names)}/{len(df_fp32_all)} shown)\n"
                 f"○ FP32  □ INT8   — — Pareto front (blue=FP32, red=INT8)")

    marker_handles = [
        plt.Line2D([0], [0], marker="o", color="none", markerfacecolor="gray",
                   markeredgecolor="white", markersize=10, label="FP32"),
        plt.Line2D([0], [0], marker="s", color="none", markerfacecolor="gray",
                   markeredgecolor="white", markersize=10, label="INT8"),
    ]
    marker_legend = ax.legend(handles=marker_handles, loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=9, title="Quantization")
    ax.add_artist(marker_legend)
    ax.legend(handles=group_legend(pd.concat([df_fp32["model_name"], df_int8["model_name"]])),
              loc="upper left", bbox_to_anchor=(1.01, 0.82), fontsize=9, title="Model family")

    plt.tight_layout()
    plt.savefig(FIGURES_DIR / "pcad_accuracy_vs_size.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print("[skip] Missing accuracy or size columns")
````

### 4. Accuracy vs Compute (MACs)

````python
if not df_data.empty and all(c in df_data.columns for c in ["macs", "fp32_top1"]) and df_data["macs"].notna().any():
    df_macs = df_data[df_data["macs"].notna() & df_data["fp32_top1"].notna()].copy()
    df_macs["macs_M"] = df_macs["macs"] / 1e6

    fig, ax = plt.subplots(figsize=(12, 8))
    sizes = (df_macs["params_m"].fillna(1) * 50).clip(40, 600)

    ax.scatter(df_macs["macs_M"], df_macs["fp32_top1"],
               c=group_colors_for(df_macs["model_name"]), s=sizes,
               alpha=0.85, edgecolors="white", lw=0.5)

    for _, row in df_macs.iterrows():
        ax.annotate(row["model_name"], (row["macs_M"], row["fp32_top1"]),
                    xytext=(5, 4), textcoords="offset points", fontsize=8)

    ax.set_xscale("log")
    ax.set_xlabel("MACs  (millions, log scale)")
    ax.set_ylabel("FP32 Top-1 Accuracy (%)")
    ax.set_title("PCAD: Accuracy vs Compute  (bubble size ∝ parameter count)")
    ax.legend(handles=group_legend(df_macs["model_name"]), loc="lower right", fontsize=9, title="Model family")
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / "pcad_accuracy_vs_macs.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print("[skip] No MACs/accuracy data")
````

### Task 3 (H3) — Compression Headroom: does weight-clustering beat plain gzip on INT8?

````python
import math
from scripts.phase9.measure_compression import entropy_bits_per_weight, cluster_weights, _weight_tensors
from ml import disk_mb, gzip_mb

COMPRESS_MODEL = "alexnet_fire"
fire_ckpt_dir = LARGE_SCALE_DIR / COMPRESS_MODEL / "checkpoints"
spec = MODEL_REGISTRY[COMPRESS_MODEL]
fire_model = load_best_model(COMPRESS_MODEL, spec["ctor"], fire_ckpt_dir, torch.device("cpu"))

n_weights = sum(w.numel() for w in _weight_tensors(fire_model))
entropy = entropy_bits_per_weight(fire_model)
nominal_mb = n_weights * 8 / 8 / (1024 ** 2)
entropy_mb = n_weights * entropy / 8 / (1024 ** 2)

rows = [("INT8 nominal", 8.0, nominal_mb), ("INT8 entropy (actual)", entropy, entropy_mb)]
for k in (16, 32, 64):
    _, codebook_mb, index_mb = cluster_weights(fire_model, k)
    bits = max(1, math.ceil(math.log2(k)))
    rows.append((f"k-means k={k} ({bits}-bit)", bits, codebook_mb + index_mb))

df_compress = pd.DataFrame(rows, columns=["method", "bits_per_weight", "size_mb"])
display(df_compress)

int8_path = fire_ckpt_dir / f"qat_{COMPRESS_MODEL}.pth"
raw_mb, gz_mb = disk_mb(int8_path), gzip_mb(int8_path)
if raw_mb:
    print(f"On-disk INT8 checkpoint ({int8_path.name}): raw {raw_mb:.4f} MB -> "
          f"gzip {gz_mb:.4f} MB (ratio {raw_mb / gz_mb:.2f}x)")
````

````python
fig, ax = plt.subplots(figsize=(8, 5))
bar_colors = [NEUTRAL] + group_colors_for([COMPRESS_MODEL]) * (len(df_compress) - 1)
ax.bar(df_compress["method"], df_compress["size_mb"], color=bar_colors)
if raw_mb:
    ax.axhline(gz_mb, color="black", linestyle="--", label=f"real gzip on-disk ({gz_mb:.3f} MB)")
    ax.legend()
ax.set_ylabel("Size (MB, weights-only estimators)")
ax.set_title(f"{COMPRESS_MODEL}: compression method comparison")
plt.xticks(rotation=30, ha="right")
plt.tight_layout()
plt.savefig(FIGURES_DIR / "phase9_h3_compression_comparison.png", dpi=150)
plt.show()
````

**H3 acceptance criterion met:** every cluster count (16/32/64) beats the real gzip ratio -- gzip's DEFLATE only captures ~1.18x on top of INT8, while explicit weight-sharing has headroom up to ~2.2x (16 clusters, weights-only). Per `docs/plans/PHASE9_PLAN.md` D6, this is a measurement-only result -- no changes to `ml/checkpoint.py` in this phase.

### Task 3b — Does Clustering Actually Hold Up? Real Post-Clustering Accuracy Across the Pareto-Frontier Models

H3 above only measured *size* (does clustering beat gzip). It never checked whether the clustered weights still work. Here we run `scripts/phase9/measure_compression.py --evaluate` (real `Trainer.evaluate()` after snapping every Conv2d/Linear weight to its nearest k-means centroid, not just sizing the result) across every model on the Phase 9 accuracy-vs-size Pareto frontier (FP32 or INT8 side) plus the bypass ablation model — the same 9-model set as the decluttered chart in section 3.

````python
COMPRESSION_DIR = project_root / "results" / "phase_9_bypass_ablation"  # curated tracked copies -- outputs/local/aggregates/ is gitignored and empty on this machine

rows = []
for p in sorted(COMPRESSION_DIR.glob("*_compression_accuracy.csv")):
    d = pd.read_csv(p)
    d["model_name"] = p.stem.removesuffix("_compression_accuracy")
    rows.append(d)

if rows:
    df_comp_acc = pd.concat(rows, ignore_index=True)

    fig, ax = plt.subplots(figsize=(11, 8))
    last_points = []
    for name, g in df_comp_acc.groupby("model_name"):
        g = g[g["cluster_k"].notna()].sort_values("bits_per_weight")
        if g.empty:
            continue
        color = group_colors_for([name])[0]
        ax.plot(g["bits_per_weight"], g["top1_drop_pp"], marker="o", color=color, lw=1.8, alpha=0.9, ms=7, zorder=3)
        last = g.iloc[-1]
        last_points.append([name, color, float(last["bits_per_weight"]), float(last["top1_drop_pp"])])

    # Greedy vertical declutter: most models converge near-zero drop at 6-bit,
    # which is exactly the region we most need to read clearly.
    last_points.sort(key=lambda p: p[3])
    CLUSTER_GAP = 3.2
    label_y = []
    for _, _, _, y in last_points:
        y_adj = y if not label_y else max(y, label_y[-1] + CLUSTER_GAP)
        label_y.append(y_adj)

    for (name, color, x, y), y_label in zip(last_points, label_y):
        ax.annotate(name, (x, y_label), xytext=(10, 0), textcoords="offset points",
                    fontsize=8.5, fontweight="bold", color=color, va="center", zorder=4)
        if abs(y_label - y) > 0.3:
            ax.plot([x, x + 0.35], [y, y_label], color=color, lw=0.6, alpha=0.5, zorder=2)

    ax.axhline(0, color="black", lw=1, alpha=0.5, ls=":")
    ax.set_xticks([4, 5, 6])
    ax.set_xticklabels(["4-bit\n(k=16)", "5-bit\n(k=32)", "6-bit\n(k=64)"])
    ax.set_xlabel("k-means weight-clustering bit-width")
    ax.set_ylabel("Top-1 accuracy drop vs FP32 baseline (pp)  — lower is better")
    ax.set_title(f"PCAD Phase 9: Does weight-clustering compression preserve accuracy?  ({df_comp_acc['model_name'].nunique()} models)\n"
                 f"Pareto-frontier + ablation models, real Trainer.evaluate() post-clustering")
    ax.legend(handles=group_legend(df_comp_acc["model_name"].unique()), loc="upper right", fontsize=9, title="Model family")
    ax.set_xlim(3.7, 8.0)
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / "phase9_compression_accuracy_by_model.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print(f"[skip] No *_compression_accuracy.csv files found under {COMPRESSION_DIR} -- "
          f"run scripts/phase9/measure_compression.py --evaluate for each model first")
````

**Reading this chart:** real post-clustering accuracy (not just size) across all 9 Pareto-frontier + ablation models. At 6-bit (k=64), every AlexNet Fire/Bottleneck-family model plus `resnet18tv` and `vgg_style` loses under ~1.7pp — weight-clustering is close to free there, and (per Task 3's size numbers) still beats plain gzip. `mobilenetv2` is the clear outlier: it collapses at 4-bit/5-bit and never recovers, still down ~31pp at 6-bit — depthwise-separable convs don't tolerate this compression method the way the Fire/Bottleneck family and standard (Conv-heavy) architectures do. Aggressive settings (4-bit, 5-bit) cost real accuracy across almost every model and aren't worth the extra size savings on their own.

### Task 3c — Accuracy vs Size: k-means Clustering vs the Real INT8 QAT Pipeline

Same accuracy-vs-size framing as section 3, restricted to these 9 models, but now with a third axis: where does naive post-hoc k-means weight-clustering (no fine-tuning) land relative to the project's actual FP32→QAT→INT8 pipeline? `fp32_size_mb`/`int8_size_mb` are real serialized checkpoint sizes; the k-means points are a **weights-only theoretical estimate** (codebook + index-stream accounting, no actual bitstream/entropy coding, no int8 execution) — so treat the k-means x-position as a size *ceiling*, not a directly comparable artifact.

````python
if not df_comp_acc.empty and not df_data.empty:
    models = sorted(df_comp_acc["model_name"].unique())
    df_sub = df_data.set_index("model_name")

    fig, ax = plt.subplots(figsize=(11, 8))
    label_points = []

    for name in models:
        color = group_colors_for([name])[0]
        fp32_size = df_sub.loc[name, "fp32_size_mb"]
        int8_size = df_sub.loc[name, "int8_size_mb"]
        int8_top1 = df_sub.loc[name, "int8_top1"]

        g = df_comp_acc[df_comp_acc["model_name"] == name]
        baseline_row = g[g["cluster_k"].isna()]
        fp32_top1 = baseline_row["top1"].iloc[0] if not baseline_row.empty else df_sub.loc[name, "fp32_top1"]
        k64_row = g[g["cluster_k"] == 64].iloc[0]
        k64_size, k64_top1 = k64_row["size_mb"], k64_row["top1"]

        # Both paths colored by model family (not gray for INT8) -- makes a model's
        # two compression paths easy to trace by eye; line style still tells them apart.
        # 4-bit/5-bit k-means settings are dropped entirely -- too little accuracy left
        # to be worth showing, and they were the main source of clutter.
        ax.plot([fp32_size, int8_size], [fp32_top1, int8_top1], color=color, lw=1.6, alpha=0.75, zorder=2, ls="--")
        ax.plot([fp32_size, k64_size], [fp32_top1, k64_top1], color=color, lw=1.6, alpha=0.75, zorder=2, ls="-")

        ax.scatter([fp32_size], [fp32_top1], color=color, marker="o", s=140, zorder=4, edgecolors="white", lw=0.6)
        ax.scatter([int8_size], [int8_top1], color=color, marker="s", s=140, zorder=4, edgecolors="white", lw=0.6)
        ax.scatter([k64_size], [k64_top1], color=color, marker="^", s=110, zorder=4, edgecolors="white", lw=0.6)

        label_points.append([name, color, fp32_size, fp32_top1])

    # Greedy vertical declutter for the FP32 labels -- several models cluster around
    # x=4-8 MB / y=48-51%.
    label_points.sort(key=lambda p: p[3])
    GAP = 2.6
    label_y = []
    for _, _, _, y in label_points:
        y_adj = y if not label_y else max(y, label_y[-1] + GAP)
        label_y.append(y_adj)

    for (name, color, x, y), y_lab in zip(label_points, label_y):
        ax.annotate(name, (x, y_lab), xytext=(8, 0), textcoords="offset points",
                    fontsize=8.5, fontweight="bold", color=color, va="center", zorder=5)
        if abs(y_lab - y) > 0.3:
            ax.plot([x, x * 1.15], [y, y_lab], color=color, lw=0.5, alpha=0.5, zorder=2)

    marker_handles = [
        plt.Line2D([0], [0], marker="o", color="none", markerfacecolor="gray", markeredgecolor="white", markersize=10, label="FP32 (real checkpoint)"),
        plt.Line2D([0], [0], marker="s", color="gray", markerfacecolor="gray", markeredgecolor="white", markersize=10, ls="--", label="INT8 QAT (real checkpoint)"),
        plt.Line2D([0], [0], marker="^", color="gray", markerfacecolor="gray", markeredgecolor="white", markersize=9, ls="-", label="k-means 6-bit / k=64 (weights-only est.)"),
    ]
    marker_legend = ax.legend(handles=marker_handles, loc="upper left", bbox_to_anchor=(1.01, 1), fontsize=9, title="Compression method")
    ax.add_artist(marker_legend)
    ax.legend(handles=group_legend(models), loc="upper left", bbox_to_anchor=(1.01, 0.62), fontsize=9, title="Model family")

    ax.set_xscale("log")
    ax.set_xlabel("Model size (MB, log scale)")
    ax.set_ylabel("Top-1 Accuracy (%)")
    ax.set_title("PCAD Phase 9: Accuracy vs Size -- INT8 QAT vs k-means weight-clustering (best setting only)\n"
                 "dashed = real INT8 QAT;  solid = k-means 6-bit (k=64), same FP32 start, colored by model family")
    plt.tight_layout()
    plt.savefig(FIGURES_DIR / "phase9_compression_vs_int8_size.png", dpi=150, bbox_inches="tight")
    plt.show()
else:
    print("[skip] Need both df_comp_acc (compression CSVs) and df_data (summary JSONs) populated")
````

**How different is k-means clustering from the real INT8 QAT pipeline?** They aren't close — INT8 QAT wins on both axes for all 9 models:

| | INT8 QAT (real pipeline) | k-means 6-bit (best clustering setting) |
|---|---|---|
| **Mechanism** | Fine-tuned: ~20 epochs of fake-quantized QAT adapts weights to quantization noise before real int8 conversion | Post-hoc only: k-means applied directly to the already-converged FP32 weights, zero fine-tuning |
| **Numeric format** | Real int8 execution via fbgemm (actual integer arithmetic, CPU-optimized kernels) | Still FP32 values, just fewer distinct ones (weight-sharing) — no inference speedup without custom kernels |
| **Size** | ~35-45% smaller (real serialized checkpoint, includes biases/BN/metadata) | Larger, and it's a weights-only theoretical estimate, not an actual compressed artifact |
| **Accuracy drop** | Often negative (accuracy *improves*: `alexnet_bottleneck` -0.72pp, `resnet18tv` -4.05pp, `vgg_style` -2.14pp) or under 1pp | Never improves; ranges 0.28-4.7pp for the Fire/Bottleneck/VGG/ResNet family, and a catastrophic 31.08pp for `mobilenetv2` |

Concretely: for every model here, the real INT8 QAT checkpoint is both smaller *and* more accurate than the best k-means result. The gap isn't the compression method itself (weight-sharing vs. integer quantization) so much as QAT's fine-tuning step — it lets the model adapt to the precision loss, while k-means here is a single post-hoc snap with no adaptation at all. If k-means weight-sharing is ever wired into the real pipeline (per Task 3's D6 — not done in this phase), doing it *after* INT8 QAT with a short fine-tune pass would be the natural next step, not as a standalone FP32-only replacement for it.
